```python
from math import sqrt

def wci (x):
    v = x["WindGustSpeed"]
    minTemp	= x["MinTemp"]
    return ((10*sqrt(v)-v+10.5)*33- minTemp)
```

```python
data["WCI"] = data.apply(wci,axis = 1)
```
